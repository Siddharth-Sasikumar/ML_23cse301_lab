# Question 1 – Classification Tree: Age and Income

Build a classification tree to predict whether a person buys a house using **age** and **income**.

```python
#ch.sc.u4cse24145
import pandas as pd
from sklearn.tree import DecisionTreeClassifier, plot_tree
import matplotlib.pyplot as plt

data = pd.DataFrame({
    'Age': [23, 26, 29, 33, 36, 39, 43, 46, 51, 56, 61, 66],
    'Income': [26, 31, 36, 41, 46, 51, 56, 61, 66, 71, 76, 81],
    'Buys_House': [0, 0, 1, 1, 1, 0, 1, 1, 1, 1, 1, 1]
})

data
```

### Train the classification tree

The first split is based on **Age**. The tree then uses **Income** for further splitting.

```python
X = data[['Age', 'Income']]
y = data['Buys_House']

model = DecisionTreeClassifier(max_depth=2, criterion='gini', random_state=42)
model.fit(X, y)

print("Tree trained successfully.")
print("Root feature:", X.columns[model.tree_.feature[0]])
```

```python
plt.figure(figsize=(12, 7))
plot_tree(
    model,
    feature_names=['Age', 'Income'],
    class_names=['Does not buy', 'Buys'],
    filled=True
)
plt.show()
```

### Interpretation

- **Age** becomes the first decision node.
- The data is split according to whether age is above or below the learned threshold.
- **Income** is then considered for the next split.
- The leaf nodes give the final prediction: **Buys House** or **Does Not Buy House**.
