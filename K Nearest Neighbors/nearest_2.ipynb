# Exercise 2 – Euclidean Distance

Calculate the distance between a new entry and the existing values using the Euclidean distance formula.

### Given dataset

| Brightness | Saturation | Class |
|---:|---:|---|
| 40 | 20 | Red |
| 50 | 50 | Blue |
| 60 | 90 | Blue |
| 10 | 25 | Red |
| 70 | 70 | Blue |
| 60 | 10 | Red |
| 25 | 80 | Blue |

For demonstration, the new entry is **(45, 40)**.

### Formula

$$d = \\sqrt{(x_2-x_1)^2 + (y_2-y_1)^2}$$

```python
#ch.sc.u4cse24145
import pandas as pd
import numpy as np

# Given dataset
data = pd.DataFrame({
    'Brightness': [40, 50, 60, 10, 70, 60, 25],
    'Saturation': [20, 50, 90, 25, 70, 10, 80],
    'Class': ['Red', 'Blue', 'Blue', 'Red', 'Blue', 'Red', 'Blue']
})

print('Given Dataset:')
display(data)
```

## New entry

We use **Brightness = 45** and **Saturation = 40** as the new entry.

```python
new_brightness = 45
new_saturation = 40

data['Distance'] = np.sqrt(
    (data['Brightness'] - new_brightness) ** 2 +
    (data['Saturation'] - new_saturation) ** 2
)

print(f'New entry: ({new_brightness}, {new_saturation})\n')
print(data.to_string(index=False))
```

## Manual calculation for the first point

For **(40, 20)**:

$$d = \\sqrt{(45-40)^2 + (40-20)^2}$$

$$d = \\sqrt{25 + 400} = \\sqrt{425} \\approx 20.62$$

```python
# Display the distance calculation for every point
for _, row in data.iterrows():
    b = row['Brightness']
    s = row['Saturation']
    d = row['Distance']
    print(f'({b}, {s}) - {row["Class"]}: {d:.2f}')
```

```python
# Sort points from nearest to farthest
result = data.sort_values('Distance').reset_index(drop=True)
print('Distances from nearest to farthest:')
display(result)
```

## Optional kNN classification

Using **k = 3**, the three nearest points are selected and majority voting determines the predicted class.

```python
k = 3
nearest = result.head(k)
predicted_class = nearest['Class'].mode()[0]

print('Three nearest neighbors:')
display(nearest)
print(f'Predicted class for ({new_brightness}, {new_saturation}): {predicted_class}')
```

## Conclusion

Euclidean distance was calculated between the new entry **(45, 40)** and every existing Brightness/Saturation point. The smaller the distance, the more similar the point is to the new entry.
